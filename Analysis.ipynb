```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

def draw_triangle_with_base(ax, x_center=0.5, y_center=0.3, base_len=0.2, side_len=0.13,
                            box_size=0.33, x_shift=0.0, y_shift=0.0, rotation=0.0):
    """
    Draw an isosceles triangle (two equal sides, base user-defined) in a fixed square box,
    with optional position shift and rotation.

    Parameters:
    - ax: matplotlib axis
    - x_center, y_center: center of the fixed square box (axes coords)
    - base_len: length of triangle's base
    - side_len: length of the two equal sides
    - box_size: size of the square box (axes fraction)
    - x_shift, y_shift: shift triangle inside the square (axes fraction)
    - rotation: rotation angle in degrees (clockwise) around triangle center
    """
    # Compute triangle center position
    tri_x = x_center + x_shift
    tri_y = y_center + y_shift

    # Half base
    half_base = base_len / 2

    # Triangle vertices before rotation
    A = np.array([-half_base, 0])  # left base
    B = np.array([half_base, 0])   # right base
    height = np.sqrt(side_len**2 - half_base**2)
    C = np.array([0, height])      # apex

    triangle = np.array([A, B, C])

    # Apply rotation
    theta = np.deg2rad(rotation)
    rotation_matrix = np.array([[np.cos(theta), -np.sin(theta)],
                                 [np.sin(theta),  np.cos(theta)]])
    triangle_rotated = triangle @ rotation_matrix.T

    # Shift triangle to its position in axes coords
    triangle_rotated += np.array([tri_x, tri_y])

    # Plot triangle
    colors = ['orange', 'red', 'blue']
    for i in range(3):
        x0, y0 = triangle_rotated[i]
        x1, y1 = triangle_rotated[(i + 1) % 3]
        ax.plot([x0, x1], [y0, y1], color='k', ls='-', transform=ax.transAxes,linewidth=1)
        ax.plot([x0, x1], [y0, y1], color=colors[i], ls='--', transform=ax.transAxes,linewidth=1.5)

    # Add side labels at midpoints
#     for i in range(3):
#         mid = (triangle_rotated[i] + triangle_rotated[(i + 1) % 3]) / 2
#         ax.text(mid[0], mid[1], r'$\ell$', fontsize=14, ha='center', va='center', transform=ax.transAxes)

    # Draw fixed square box
    box_x = x_center - box_size / 2
    box_y = y_center - box_size / 2
    box = patches.Rectangle((box_x, box_y), box_size, box_size,
                            linewidth=1, edgecolor='black', facecolor='none', transform=ax.transAxes)
    ax.add_patch(box)
```

```python
import numpy as np
import pandas as pd

# -----------------------------------------------------------------------------
# 1) Don’t blank out *all* duplicates up front.  Instead we’ll blank only within
#    each *continuous* run of rows.  That way at a page‐break you’ll see the
#    first k1 of that page.
# -----------------------------------------------------------------------------
df_plot = df_fmt.copy()

# mark where k1 changes
is_new = (df_plot['k1'] != df_plot['k1'].shift(1))

# build a column that *only* blanks when k1 == previous k1
df_plot['k1_disp'] = np.where(is_new, df_plot['k1'], "")

# -----------------------------------------------------------------------------
# 2) Grab the raw rows from pandas (as before)
# -----------------------------------------------------------------------------
raw = (
    df_plot[['k1_disp','mu','t','BS','1sig','UL']]
      .to_latex(index=False, header=False, escape=False)
      .splitlines()
)
body = raw[2:-2]

# -----------------------------------------------------------------------------
# 3) Build your longtable, making sure to put \endfirsthead immediately after
#    the *one* header block, and to *not* emit a \caption[]{…} in that continued
#    header (so it won’t re‐caption—and thus re‐header—on page 1).
# -----------------------------------------------------------------------------
lines = []

lines.append(r"\begin{longtable}{l ll ccc}")
lines.append(r"\caption{Bispectrum measurements}\label{tab:bispec}\\")
lines.append(r"\toprule")
lines.append(r"\multicolumn{1}{c}{\textbf{Size}} & "
             r"\multicolumn{2}{c}{\textbf{Shape}} & "
             r"\multicolumn{3}{c}{\textbf{Bispectrum}} \\")
lines.append(r"\cmidrule(l){1-1} \cmidrule(l){2-3} \cmidrule(l){4-6}")
lines.append(r"$k_1$\,[Mpc$^{-1}$] & $\mu$ & $t$ & $\Delta^3$\,[mK$^{6}$] & $1\sigma$\,[mK$^{6}$] & $\Delta^3_{\rm UL}$\,[mK$^{6}$] \\")
lines.append(r"\midrule")
lines.append(r"\endfirsthead")

# continued header for pages 2,3,… (no \caption here!)
lines.append(r"\toprule")
lines.append(r"\multicolumn{1}{c}{\textbf{Size}} & "
             r"\multicolumn{2}{c}{\textbf{Shape}} & "
             r"\multicolumn{3}{c}{\textbf{Bispectrum}} \\")
lines.append(r"\cmidrule(l){1-1} \cmidrule(l){2-3} \cmidrule(l){4-6}")
lines.append(r"$k_1$\,[Mpc$^{-1}$] & $\mu$ & $t$ & $\Delta^3$\,[mK$^{6}$] & $1\sigma$\,[mK$^{6}$] & $\Delta^3_{\rm UL}$\,[mK$^{6}$] \\")
lines.append(r"\midrule")
lines.append(r"\endhead")

# footer on all but last page
lines.append(r"\midrule")
lines.append(r"\multicolumn{6}{r}{\textit{continued on next page}} \\")
lines.append(r"\endfoot")

# footer on last page
lines.append(r"\bottomrule")
lines.append(r"\endlastfoot")

# -----------------------------------------------------------------------------
# 4) Finally dump in your body rows, inserting our mid‐rules at each new k1.
# -----------------------------------------------------------------------------
row_i = 0
for idx, new_flag in zip(df_plot.index, is_new):
    if new_flag and row_i != 0:
        lines.append(r"\midrule")
    lines.append(body[row_i])
    row_i += 1

lines.append(r"\end{longtable}")

print("\n".join(lines))
```
