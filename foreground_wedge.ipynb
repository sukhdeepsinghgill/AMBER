```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

def draw_triangle(fig, k1, k2, theta, loc=(0.1, 0.1), size=0.2):
    """
    Draw a triangle with sides k1, k2 and included angle theta, on fig as an inset.

    Parameters
    ----------
    fig : matplotlib.figure.Figure
        The figure to draw into.
    k1, k2 : float
        Lengths of the two user-defined sides (k_{1⊥} and k_{2⊥}).
    theta : float
        Included angle between k1 and k2 (in radians).
    loc : tuple(float, float), optional
        (x0, y0) lower-left corner of the inset axes, in figure fraction [0,1].
    size : float, optional
        Width (and height) of the inset axes, in figure fraction [0,1].

    Notes
    -----
    - Computes k3 via:  
      k3 = sqrt(k1^2 + k2^2 − 2 k1 k2 cos(theta))
    - Labels sides k_{1⊥}, k_{2⊥}, k_{3⊥}, and the angle θ at the shared vertex.
    """
    # compute the third side
    k3 = np.sqrt(k1**2 + k2**2 - 2*k1*k2*np.cos(theta))

    # create inset axes
    ax = fig.add_axes([loc[0], loc[1], size, size])
    ax.axis('off')

    # define triangle in “real” coordinates
    P0 = np.array([0.0, 0.0])
    P1 = np.array([k1, 0.0])
    P2 = np.array([k2 * np.cos(theta), k2 * np.sin(theta)])
    pts = np.vstack([P0, P1, P2])

    # normalize into [0,1]×[0,1]
    mn, mx = pts.min(), pts.max()
    pts = (pts - mn) / (mx - mn)

    # draw triangle
    tri = Polygon(pts, fill=False, edgecolor='black', linewidth=1.5)
    ax.add_patch(tri)
    ax.set_xlim(-0.05, 1.05)
    ax.set_ylim(-0.05, 1.05)
    
    A, B, C = pts

    # draw each side separately with its color
    ax.plot([A[0], B[0]], [A[1], B[1]], color='r', lw=3,linestyle='--')
    ax.plot([A[0], C[0]], [A[1], C[1]], color='g', lw=3,linestyle='--')
    ax.plot([B[0], C[0]], [B[1], C[1]], color='b', lw=3,linestyle='--')

    # label mids
    midAB = (A+B)/2
    midAC = (A+C)/2
    midBC = (B+C)/2
    ax.text(*midAB+[-.025,-.04], r'$k_{1\perp}$', ha='center', va='top',fontsize=18)
    ax.text(*midAC+[-.02,.02], r'$k_{3\perp}$', ha='right', va='center',rotation=80,fontsize=18)
    ax.text(*midBC+[.02,.02], r'$k_{2\perp}$', ha='center', va='bottom',rotation=-20,fontsize=18)


    # angle label at P0 in axes‐fraction coords
    ax.text(B[0]-.25, B[1]+0.06, r'$\theta$', transform=ax.transAxes,
            ha='right', va='bottom',fontsize=18)

    return ax
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ─── USER PARAMETERS ───────────────────────────────────────────────────────────
r        = 9200.0   # comoving distance to the source (e.g. in Mpc)
r_prime  = 17.0   # dr_parallel / dnu scaling (e.g. in Mpc/MHz)
theta0   = np.deg2rad(90)   # angular offset of the point source (radians)
nu_c     = 154.255   # central observing frequency (MHz)
S        = 1.0   # source flux (arbitrary units)
# ────────────────────────────────────────────────────────────────────────────────

# Build (k_perp, k_parallel) grid in polar coords
k_perp_vals     = np.linspace(0, .2, 500)
k_parallel_vals = np.linspace(0, 5, 500)
phi_vals        = np.linspace(0, 2*np.pi, 180)  # azimuthal angle in k_perp plane

KP, KPAR = np.meshgrid(k_perp_vals, k_parallel_vals)

# We integrate over phi (direction of K_perp)
P = np.zeros_like(KP)

for phi in phi_vals:
    proj = (r * KP * np.cos(phi) * theta0) / nu_c  # projection of K_perp ⋅ theta
    step = (proj > r_prime * KPAR).astype(float)
    P += step

# Normalize over phi
P = S**2 * P / len(phi_vals)

# Plot
plt.figure(figsize=(6, 5))
plt.imshow(
    P,
    extent=(k_perp_vals.min(), k_perp_vals.max(), k_parallel_vals.min(), k_parallel_vals.max()),
    origin='lower',
    aspect='auto',
    cmap='viridis'
)
plt.colorbar(label=r'$P(k_\perp, k_\parallel)$')
plt.xlabel(r'$k_\perp$')
plt.ylabel(r'$k_\parallel$')
plt.title('Power Spectrum $P(k_\perp, k_\parallel)$ for Point Source at $\\theta_0$')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ── USER PARAMETERS ─────────────────────────────────────────────────────────
r        = 9200.0    # comoving distance [Mpc]
r_prime  = 17.0    # dr_parallel/dnu [Mpc/MHz]
nu_c     = 154.255    # central freq [MHz]
theta    = [0.2,  0.5, -0.3]  # three source angles [rad]
S        = [1.0,  1,  1]  # their fluxes
sigma    = 0.02   # Gaussian width for δ‐approx.
# ────────────────────────────────────────────────────────────────────────────

# compute the slopes m_i = r*theta_i/(r'*nu_c)
m = [r*th/(r_prime*nu_c) for th in theta]

# set up a 2D grid in (k1_perp, k2_perp)
k1 = np.linspace(0,5,400)
k2 = np.linspace(0,5,400)
K1, K2 = np.meshgrid(k1, k2)

# enforce triangle: k3 = -(k1 + k2); and parallel modes kpar_i = m_i * ki_perp
K3   = -(K1 + K2)
kp1  = m[0]*K1
kp2  = m[1]*K2
kp3  = m[2]*K3

# approximate each δ_D by exp[−(r k_perp θ/ν_c − r' k_parallel)^2/(2σ^2)]
G1 = np.exp(-((r*K1*theta[0]/nu_c - r_prime*kp1)**2)/(2*sigma**2))
G2 = np.exp(-((r*K2*theta[1]/nu_c - r_prime*kp2)**2)/(2*sigma**2))
G3 = np.exp(-((r*K3*theta[2]/nu_c - r_prime*kp3)**2)/(2*sigma**2))

# bispectrum amplitude on this grid
B = S[0]*S[1]*S[2] * G1 * G2 * G3

# plot a slice, e.g. at k2_perp = 0:
plt.figure(figsize=(6,4))
plt.plot(k1, B[B.shape[0]//2,:])
plt.xlabel(r'$k_{1\perp}$ (with $k_{2\perp}=0$)')
plt.ylabel(r'$B(k_{1},k_{2}=0)$')
plt.title('Bispectrum Slice for Three Point Sources')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ─── PARAMETERS ────────────────────────────────────────────────────────────────
r = 9200.0          # Mpc
r_prime = 17.0    # Mpc/Hz
nu_c = 154.255      # Hz
S = 1.0          # Arbitrary units

# Source angles (radians)
theta1 = 0.3
theta2 = 0.5
theta3 = 0.7

# Fixed magnitude of k_perp
k_perp = 1e-2

# k_parallel range
k_par_vals = np.linspace(-4, 4, 500)
phi_vals = np.linspace(0, 2*np.pi, 180)

# Initialize bispectrum array
B = np.zeros((len(k_par_vals), len(phi_vals)))

for i, k_par in enumerate(k_par_vals):
    k1_par = k_par
    k2_par = k_par
    k3_par = -2 * k_par  # due to constraint

    for j, phi in enumerate(phi_vals):
        # Build triangle configuration in k_perp
        K1 = k_perp * np.array([np.cos(phi), np.sin(phi)])
        K2 = k_perp * np.array([np.cos(phi + 2*np.pi/3), np.sin(phi + 2*np.pi/3)])
        K3 = -K1 - K2

        # Project onto theta vectors
        V1 = (r * np.dot(K1, [theta1, 0])) / nu_c - r_prime * k1_par
        V2 = (r * np.dot(K2, [theta2, 0])) / nu_c - r_prime * k2_par
        V3 = (r * np.dot(K3, [theta3, 0])) / nu_c + r_prime * (k1_par + k2_par)

        # Multiply theta functions (Heaviside step)
        B[i, j] = S**3 * (V1 > 0) * (V2 > 0) * (V3 > 0)

# Average over angle phi to get 1D wedge-like bispectrum
B_avg = np.mean(B, axis=1)

# Plot
plt.figure(figsize=(6, 5))
plt.plot(k_par_vals, B_avg, color='darkred')
plt.xlabel(r'$k_\parallel$')
plt.ylabel(r'Bispectrum Amplitude $B(k_\perp, k_\parallel)$')
plt.title('Foreground Wedge in Bispectrum (Averaged over Triangle Orientation)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ─── PARAMETERS ────────────────────────────────────────────────────────────────
r = 9200.0          # Mpc
r_prime = 17.0    # Mpc/Hz
nu_c = 154.255       # Hz
S = 1.0          # Source flux (arbitrary units)

# Angular positions (in radians) of three point sources
theta1 = 0.3
theta2 = 0.5
theta3 = 0.7

# Grid definitions
k_perp_vals = np.linspace(0, 1, 100)     # Magnitudes of k_perp
k_par_vals  = np.linspace(-5, 5, 100)     # k_parallel values
phi_vals    = np.linspace(0, 2*np.pi, 90) # Orientation angles in the transverse plane

# Bispectrum array
B = np.zeros((len(k_par_vals), len(k_perp_vals)))

# Loop over k_parallel and k_perp
for i, k_par in enumerate(k_par_vals):
    k1_par = k_par
    k2_par = k_par
    k3_par = -2 * k_par  # Constraint: k3_parallel = -k1_parallel - k2_parallel

    for j, kp in enumerate(k_perp_vals):
        # Sum over triangle orientations
        prod_sum = 0
        for phi in phi_vals:
            # Triangle configuration in k_perp space
            K1 = kp * np.array([np.cos(phi), np.sin(phi)])
            K2 = kp * np.array([np.cos(phi + 2*np.pi/3), np.sin(phi + 2*np.pi/3)])
            K3 = -K1 - K2  # Enforce triangle closure

            # Compute visibility step arguments
            V1 = (r * np.dot(K1, [theta1, 0])) / nu_c - r_prime * k1_par
            V2 = (r * np.dot(K2, [theta2, 0])) / nu_c - r_prime * k2_par
            V3 = (r * np.dot(K3, [theta3, 0])) / nu_c - r_prime * k3_par

            # Multiply theta functions (step functions)
            prod_sum += (V1 > 0) * (V2 > 0) * (V3 > 0)

        # Average over phi and scale
        B[i, j] = (S**3) * (prod_sum / len(phi_vals))

# ─── PLOT ───────────────────────────────────────────────────────────────────────
plt.figure(figsize=(6, 5))
plt.imshow(
    B,
    extent=(k_perp_vals.min(), k_perp_vals.max(), k_par_vals.min(), k_par_vals.max()),
    origin='lower',
    aspect='auto'
)
plt.colorbar(label=r'$B(k_\perp, k_\parallel)$')
plt.xlabel(r'$k_\perp$')
plt.ylabel(r'$k_\parallel$')
plt.title('Bispectrum Foreground Wedge in $(k_\perp, k_\parallel)$ Space')
plt.tight_layout()
plt.show()
```
