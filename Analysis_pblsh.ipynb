```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

def draw_triangle(fig, k1, k2, theta, loc=(0.1, 0.1), size=0.2):
    """
    Draw a triangle with sides k1, k2 and included angle theta, on fig as an inset.

    Parameters
    ----------
    fig : matplotlib.figure.Figure
        The figure to draw into.
    k1, k2 : float
        Lengths of the two user-defined sides (k_{1⊥} and k_{2⊥}).
    theta : float
        Included angle between k1 and k2 (in radians).
    loc : tuple(float, float), optional
        (x0, y0) lower-left corner of the inset axes, in figure fraction [0,1].
    size : float, optional
        Width (and height) of the inset axes, in figure fraction [0,1].

    Notes
    -----
    - Computes k3 via:  
      k3 = sqrt(k1^2 + k2^2 − 2 k1 k2 cos(theta))
    - Labels sides k_{1⊥}, k_{2⊥}, k_{3⊥}, and the angle θ at the shared vertex.
    """
    # compute the third side
    k3 = np.sqrt(k1**2 + k2**2 - 2*k1*k2*np.cos(theta))

    # create inset axes
    ax = fig.add_axes([loc[0], loc[1], size, size])
    ax.axis('off')

    # define triangle in “real” coordinates
    P0 = np.array([0.0, 0.0])
    P1 = np.array([k1, 0.0])
    P2 = np.array([k2 * np.cos(theta), k2 * np.sin(theta)])
    pts = np.vstack([P0, P1, P2])

    # normalize into [0,1]×[0,1]
    mn, mx = pts.min(), pts.max()
    pts = (pts - mn) / (mx - mn)

    # draw triangle
    tri = Polygon(pts, fill=False, edgecolor='black', linewidth=1.5)
    ax.add_patch(tri)
    ax.set_xlim(-0.05, 1.05)
    ax.set_ylim(-0.05, 1.05)
    
    A, B, C = pts

    # draw each side separately with its color
    ax.plot([A[0], B[0]], [A[1], B[1]], color='r', lw=3,linestyle='--')
    ax.plot([A[0], C[0]], [A[1], C[1]], color='g', lw=3,linestyle='--')
    ax.plot([B[0], C[0]], [B[1], C[1]], color='b', lw=3,linestyle='--')

    # label mids
    midAB = (A+B)/2
    midAC = (A+C)/2
    midBC = (B+C)/2
    ax.text(*midAB+[-.025,-.04], r'$k_{1}$', ha='center', va='top',fontsize=18)
    ax.text(*midAC+[-.02,.02], r'$k_{3}$', ha='right', va='center',rotation=80,fontsize=18)
    ax.text(*midBC+[.02,.02], r'$k_{2}$', ha='center', va='bottom',rotation=-20,fontsize=18)


    # angle label at P0 in axes‐fraction coords
    ax.text(B[0]-.25, B[1]+0.06, r'$\theta$', transform=ax.transAxes,
            ha='right', va='bottom',fontsize=18)

    return ax
```

```python
import numpy as np
import pandas as pd

# -----------------------------------------------------------------------------
# 1) Don’t blank out *all* duplicates up front.  Instead we’ll blank only within
#    each *continuous* run of rows.  That way at a page‐break you’ll see the
#    first k1 of that page.
# -----------------------------------------------------------------------------
df_plot = df_fmt.copy()

# mark where k1 changes
is_new = (df_plot['k1'] != df_plot['k1'].shift(1))

# build a column that *only* blanks when k1 == previous k1
df_plot['k1_disp'] = np.where(is_new, df_plot['k1'], "")

# -----------------------------------------------------------------------------
# 2) Grab the raw rows from pandas (as before)
# -----------------------------------------------------------------------------
raw = (
    df_plot[['k1_disp','mu','t','BS','1sig','UL']]
      .to_latex(index=False, header=False, escape=False)
      .splitlines()
)
body = raw[2:-2]

# -----------------------------------------------------------------------------
# 3) Build your longtable, making sure to put \endfirsthead immediately after
#    the *one* header block, and to *not* emit a \caption[]{…} in that continued
#    header (so it won’t re‐caption—and thus re‐header—on page 1).
# -----------------------------------------------------------------------------
lines = []

lines.append(r"\begin{longtable}{l ll ccc}")
lines.append(r"\caption{Bispectrum measurements}\label{tab:bispec}\\")
lines.append(r"\toprule")
lines.append(r"\multicolumn{1}{c}{\textbf{Size}} & "
             r"\multicolumn{2}{c}{\textbf{Shape}} & "
             r"\multicolumn{3}{c}{\textbf{Bispectrum}} \\")
lines.append(r"\cmidrule(l){1-1} \cmidrule(l){2-3} \cmidrule(l){4-6}")
lines.append(r"$k_1$\,[Mpc$^{-1}$] & $\mu$ & $t$ & $\Delta^3$\,[mK$^{6}$] & $1\sigma$\,[mK$^{6}$] & $\Delta^3_{\rm UL}$\,[mK$^{6}$] \\")
lines.append(r"\midrule")
lines.append(r"\endfirsthead")

# continued header for pages 2,3,… (no \caption here!)
lines.append(r"\toprule")
lines.append(r"\multicolumn{1}{c}{\textbf{Size}} & "
             r"\multicolumn{2}{c}{\textbf{Shape}} & "
             r"\multicolumn{3}{c}{\textbf{Bispectrum}} \\")
lines.append(r"\cmidrule(l){1-1} \cmidrule(l){2-3} \cmidrule(l){4-6}")
lines.append(r"$k_1$\,[Mpc$^{-1}$] & $\mu$ & $t$ & $\Delta^3$\,[mK$^{6}$] & $1\sigma$\,[mK$^{6}$] & $\Delta^3_{\rm UL}$\,[mK$^{6}$] \\")
lines.append(r"\midrule")
lines.append(r"\endhead")

# footer on all but last page
lines.append(r"\midrule")
lines.append(r"\multicolumn{6}{r}{\textit{continued on next page}} \\")
lines.append(r"\endfoot")

# footer on last page
lines.append(r"\bottomrule")
lines.append(r"\endlastfoot")

# -----------------------------------------------------------------------------
# 4) Finally dump in your body rows, inserting our mid‐rules at each new k1.
# -----------------------------------------------------------------------------
row_i = 0
for idx, new_flag in zip(df_plot.index, is_new):
    if new_flag and row_i != 0:
        lines.append(r"\midrule")
    lines.append(body[row_i])
    row_i += 1

lines.append(r"\end{longtable}")

print("\n".join(lines))
```
